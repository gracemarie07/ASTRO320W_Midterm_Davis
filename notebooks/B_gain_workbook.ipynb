```python
from pathlib import Path
import os
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Image
%matplotlib inline
ROOT = Path.cwd().parent
DATA = ROOT / 'data'
OUT = ROOT / 'results'
OUT.mkdir(exist_ok=True)
```

```python
import openpyxl
workbook = next((ROOT/'data').glob('*.xlsx'))
sheet = openpyxl.load_workbook(workbook, data_only=True).active
pairs = [row[:4] for row in sheet.iter_rows(min_row=2, values_only=True)
         if row[0] is not None and all(isinstance(x, (int, float)) for x in row[1:4])]
label, f1, f2, noise = zip(*pairs)
gain = (np.array(f1) + np.array(f2)) / np.array(noise)**2
for pair, value in zip(label, gain):
    print(f'{pair}: {value:.6f} electrons/DN')
print('Mean:', gain.mean())
print('Sample SD:', gain.std(ddof=1))
print('SEM:', gain.std(ddof=1)/np.sqrt(len(gain)))
fig, ax = plt.subplots(figsize=(7,4))
ax.plot(np.arange(1,len(gain)+1), gain, 'o')
ax.axhline(gain.mean(), color='gray', linestyle='--')
ax.set(xlabel='Flat pair', ylabel='Photon-only gain (electrons/DN)')
plt.show()
```
