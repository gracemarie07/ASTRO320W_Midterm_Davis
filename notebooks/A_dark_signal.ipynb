```python
from pathlib import Path
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
out=OUT
plt.rcParams.update({'font.size':11,'axes.labelsize':11,'axes.titlesize':12,'legend.fontsize':9})
t=np.array([.1,6,10,30,45,60]);q=np.array([31.568,58.079,113.585,279.962,452.548,654.913])*1.34
coef,cov=np.polyfit(t[1:],q[1:],1,cov=True)
rate=q[1:]/t[1:]
print('Time slope and SE',coef[0],np.sqrt(cov[0,0]),'mean rate / SEM',rate.mean(),rate.std(ddof=1)/np.sqrt(5))
f,ax=plt.subplots(figsize=(6.5,4.4),layout='constrained');ax.scatter(t[1:],q[1:],label='6–60 s measurements',color='#245884');ax.scatter(t[0],q[0],marker='s',facecolors='none',edgecolors='#a45e20',label='0.1 s (excluded from fit)');x=np.linspace(0,62,200);ax.plot(x,np.polyval(coef,x),color='#245884',lw=1,label='Unweighted fit');ax.set(xlabel='Exposure time (s)',ylabel=r'Nominal charge ($e^-$ pixel$^{-1}$)',title='Accumulated dark charge versus exposure time');ax.legend();ax.grid(alpha=.2);f.savefig(out/'01_dark_charge_vs_time.png',dpi=240);plt.close(f)
T=np.array([-7.844,.844,18.750,20.219]);raw=np.array([1048.4342,1092.8636,1338.7558,1922.2905]);Q=(raw-1038.930125)*1.34;I=Q/300
f,ax=plt.subplots(figsize=(6.5,4.4),layout='constrained');ax.scatter(T,Q,color='#245884');ax.set(xlabel='Recorded starting CCD temperature (°C)',ylabel=r'Nominal charge ($e^-$ pixel$^{-1}$)',title='Accumulated dark charge versus temperature');ax.grid(alpha=.2);f.savefig(out/'02_dark_charge_vs_temperature.png',dpi=240);plt.close(f)
x=1/(T+273.15);y=np.log(I);c,v=np.polyfit(x,y,1,cov=True);k=8.617333e-5
print('Activation / SE',-k*c[0],k*np.sqrt(v[0,0]))
f,ax=plt.subplots(figsize=(6.5,4.4),layout='constrained');ax.scatter(x*1e3,y,color='#245884',label='300-s observations');xx=np.linspace(x.min(),x.max(),200);ax.plot(xx*1e3,np.polyval(c,xx),lw=1,color='#245884',label='Unweighted fit');ax.set(xlabel=r'Inverse starting temperature ($10^{-3}$ K$^{-1}$)',ylabel=r'$\ln(I/I_{\rm ref})$',title='Linearized temperature dependence');ax.legend();ax.grid(alpha=.2);f.savefig(out/'03_linearized_thermal_fit.png',dpi=240);plt.close(f)
```
